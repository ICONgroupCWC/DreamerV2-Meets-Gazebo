```python
def compute_cs_distance(channels, device):
    channels = channels.to(device)
    powers = torch.einsum("lbmt,lbmt->lbt", channels, torch.conj(channels))

    # --- FIX ZERO POWERS ---
    powers_abs = torch.abs(powers)
    powers_flat = powers_abs.flatten()

    nonzero = powers_flat[powers_flat > 0]
    if len(nonzero) == 0:
        raise ValueError("All channel powers are zero. Cannot compute CS distance.")

    min_nonzero = torch.min(nonzero)
    replacement_value = min_nonzero * 0.5

    # replace zeros
    powers_abs[powers_abs == 0] = replacement_value
    powers = powers_abs
    # ------------------------

    cs_distance = torch.zeros((channels.shape[0], channels.shape[0])).to(device)

    for ii in tqdm(range(channels.shape[0])):
        h = channels[ii]
        w = channels[ii:]
        dotproducts = torch.abs(
            torch.square(torch.einsum("bmt,lbmt->lbt", torch.conj(h), w))
        )
        d_new = torch.sum(
            torch.sum(1 - dotproducts / (powers[ii] * powers[ii:]), dim=-1),
            dim=-1
        )
        d = torch.cat((torch.zeros(ii, device=device), torch.max(d_new, torch.zeros_like(d_new))))
        cs_distance[ii] = d

    cs_distance = cs_distance + cs_distance.T
    return cs_distance
cs_distance = compute_cs_distance(channels.reshape(-1, 3, 8, 16), torch.device("cuda"))
```

```python
n_neighbors = 20

neighbors_alg = NearestNeighbors(
    n_neighbors=n_neighbors, metric="precomputed", n_jobs=-1
)
nbrs = neighbors_alg.fit(cs_distance.cpu())
nbg = kneighbors_graph(nbrs, n_neighbors, metric="precomputed", mode="distance")
```

```python
dissimilarity_matrix_geodesic = torch.zeros(
    (nbg.shape[0], nbg.shape[1]), dtype=torch.float32, device="cuda"
)

for i in tqdm(range(nbg.shape[0])):
    dissimilarity_matrix_geodesic[i, :] = torch.from_numpy(
        dijkstra(nbg, directed=False, indices=i)
    )
```

```python
import multiprocessing as mp
import queue

dissimilarity_matrix_geodesic = np.zeros((nbg.shape[0], nbg.shape[1]), dtype = np.float32)

def shortest_path_worker(todo_queue, output_queue):
	while True:
		index = todo_queue.get()
		if index == -1:
			output_queue.put((-1, None))
			break
		d = dijkstra(nbg, directed=False, indices=index)
		output_queue.put((index, d))

with tqdm(total = nbg.shape[0]**2) as pbar:
	todo_queue = mp.Queue()
	output_queue = mp.Queue()

	for i in range(nbg.shape[0]):
		todo_queue.put(i)

	for i in range(mp.cpu_count()):
		todo_queue.put(-1)
		p = mp.Process(target = shortest_path_worker, args = (todo_queue, output_queue))
		p.start()

	finished_processes = 0
	while finished_processes != mp.cpu_count():
		i, d = output_queue.get()

		if i == -1:
			finished_processes = finished_processes + 1
		else:
			dissimilarity_matrix_geodesic[i,:] = d
			pbar.update(len(d))
```

```python
class CS_SiameseLoss(nn.Module):
    def __init__(self, device, beta=1.0):
        super(CS_SiameseLoss, self).__init__()
        self.beta = beta
        self.device = device

    def forward(self, channels, embeddings, indices):

        predicted_distance = torch.cdist(embeddings, embeddings)

        loss = torch.mean(
            torch.square(
                predicted_distance
                - dissimilarity_matrix_geodesic[:, indices][indices]
            )
        )

        return loss
```

```python
from scipy.spatial import distance_matrix
import scipy

def kruskal_stress(X, X_embedded):
	dist_X = distance_matrix(X, X)
	dist_X_embedded = distance_matrix(X_embedded, X_embedded)
	beta = np.divide(np.sum(dist_X * dist_X_embedded), np.sum(dist_X_embedded * dist_X_embedded))

	return np.sqrt(np.divide(np.sum(np.square((dist_X - beta * dist_X_embedded))), np.sum(dist_X * dist_X)))

ks = kruskal_stress(locations.reshape(-1, 2), chart.cpu())
print(f"KS: {ks:.6f}")
```

```python
from utils.models import Encoder, Predictor
from utils.datasets import ChannelDataset, JEPA_ChannelDataset
from utils.training import cc_train, jepa_train

device = torch.device("cuda")
enc_lr = 0.002
pre_lr = 0.002
momentum_step = 0.99
num_epochs = 1500
batch_size = 100

seq_len = 45
window_len = 20

encoder = Encoder().to(device)
encoder.load_state_dict(torch.load("enc"))
# encoder = copy.deepcopy(cc_encoder)
predictor = Predictor(input_dim=1324, hidden_dim=256, num_layers=1, output_dim=2).to(device)
# predictor = Predictor(input_dim=300, hidden_dim=256, num_layers=1, output_dim=2).to(device)
target_encoder = copy.deepcopy(encoder)
# print(encoder)
# print(predictor)
# print(target_encoder)
param_groups = [
    {"params": (p for n, p in encoder.named_parameters() if (('bias' not in n) and (len(p.shape) != 1))), "lr": enc_lr, "weight_decay": 3e-3},
    {"params": (p for n, p in predictor.named_parameters() if (('bias' not in n) and (len(p.shape) != 1))), "lr": pre_lr, "weight_decay": 0},
    {"params": (p for n, p in encoder.named_parameters() if (('bias' in n) or (len(p.shape) == 1))), "lr": enc_lr},
    {"params": (p for n, p in predictor.named_parameters() if (('bias' in n) or (len(p.shape) == 1))), "lr": pre_lr},
]

optimizer = optim.Adam(param_groups)
scheduler = StepLR(optimizer=optimizer, step_size=1, gamma=0.97)
jepa_loss_fn = torch.nn.MSELoss().to(device)

# jepa_train_dataset = torch.utils.data.ConcatDataset([
#     JEPA_ChannelDataset(channels[ep], latent_space[ep], seq_len, window_len) for ep in range(locations.shape[0])
# ])
jepa_train_dataset = JEPA_ChannelDataset(
    channels,       # shape (12471, 3, 8, 16)
    latent_space,   # shape (12471, 1324)
    seq_len=seq_len,      # CSI has length 3 → cannot use more
    window_len=window_len
)
print("Dataset length:", len(jepa_train_dataset))

jepa_train_dataloader = DataLoader(jepa_train_dataset, batch_size=batch_size, shuffle=True, drop_last=True)

losses = {"train": [], "val": []}
for epoch in (pbar := tqdm(range(num_epochs))):
    epoch_loss = jepa_train(
        encoder,
        predictor,
        target_encoder,
        jepa_train_dataloader,
        jepa_loss_fn,
        optimizer,
        momentum_step,
        device
    )
    losses["train"].append(epoch_loss["train"])
    losses["val"].append(epoch_loss["val"])
    scheduler.step()
    pbar.set_description(f'Epoch {epoch+1}/{num_epochs}, Training Loss: {epoch_loss["train"]:.6f}, Validation Loss: {epoch_loss["val"]:.6f}')
```

```python
def affine_transform_channel_chart(groundtruth_pos, channel_chart_pos):
    pad = lambda x: np.hstack([x, np.ones((x.shape[0], 1))])
    unpad = lambda x: x[:,:-1]
    A, res, rank, s = np.linalg.lstsq(pad(channel_chart_pos), pad(groundtruth_pos), rcond = None)
    transform = lambda x: unpad(np.dot(pad(x), A))
    return transform(channel_chart_pos)

cc_positions_transformed = affine_transform_channel_chart(locations.reshape(-1, 2).numpy(), chart.numpy())
# plt.scatter(cc_positions_transformed[:, 0], cc_positions_transformed[:, 1], c=cl, s=5)
# plt.show()
```
